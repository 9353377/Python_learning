```python
""" 7. Write a Python program to find the sum of the digits of a number."""

num = int(input("enter the number :"))

total = 0

while num > 0:
    digit =  num % 10
    total = total + digit
    num = num // 10

print(total)
```

```python
""" 9. Check if a Number is a Palindrome

A number is a palindrome if it reads the same forwards and backwards."""

Onum= int(input("enter the number:"))
original=Onum
reverse=0
while Onum>0:
    Ldigit=Onum % 10
    reverse=reverse*10+Ldigit
    Onum=Onum//10
print("Reversed number=",reverse)
if reverse==original:
    print("It is a plindrome!")
else:
    print("not palindrome")
    
    
    
```

```python
""" 11. Write a program that prints the first n Fibonacci numbers."""

n= int(input("Enter the num:"))
first = 0
second= 1
count =0
while count < n:
    print(first, end=" ")
    next_n=first+second
    first=second
    second=next_n
    
    count=count +1
    
```
