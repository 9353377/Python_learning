```python
""" 8) Write a Python program to check whether a number is Prime using a for loop."""

n=int(input("ente the num: "))
if n<=1:
    print("not prime")
else:
    is_prime=True
    for i in range(2,n):
        if n%i==0:
            is_prime=False
            
            break
    if is_prime:
        print("prime")
    else:
        print("not prime")

```

```python
""" 9) Write a Python program to print the first n Fibonacci numbers using a for loop."""
n=int(input("ente the num: "))
first=0
second=1
for i in range(1,n+1):
    print(first)
    next_n=first+second
    first=second
    
    second=next_n
    
```

```python
""" 10) Write a Python program to check whether a number is an Armstrong number"""
n=int(input("ente the num: "))
total=0
original=n
for i in range(0,n):
    digit=n%10
    total=total+(digit**3)
    n=n//10
print("Total :",total)
if total== original:
    print("It is a Armstrong number")
else:
    print("Not")
```

```python
""" 22) Count the number of uppercase letters, lowercase letters, digits, and special characters in a string."""
s=input("enter the string:")
upper_count=0
l_count=0
d_count=0
s_count=0
for i in s:
    if "A" <= i <="Z":
        upper_count=upper_count+1
        
    elif "a" <= i <="z":
        l_count=l_count+1
    elif "0" <= i <="9":
        d_count=d_count+1
    else:
        s_count=s_count+1
print("upperletter",upper_count)
print("loweletter",l_count)
print("Digits",d_count)
print("specialCharacters",s_count)
    
    
    
```
