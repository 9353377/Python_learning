```python
""" 2. Write a Python program that takes a person's age as input and prints:

"Child" if age is less than 13.
"Teenager" if age is between 13 and 19 (inclusive).
"Adult" if age is 20 or above"""


age = int(input("Enter the age of person:"))
if age>0 and age < 13:
    print("child")
elif age >= 13 and age <=19:
    print("Teenager")
elif age >= 20:
    print("Adult")
else:
    print("invalid")
```

```python
"""5.Write a Python program that takes an integer as input and prints:

"Divisible by both" if the number is divisible by 3 and 5.
"Divisible by 3" if it is divisible only by 3.
"Divisible by 5" if it is divisible only by 5.
"Not divisible by either" otherwise."""

num = int(input("enter an integer : "))
if (num % 3 ==0 and num % 5==0):
    print("Divisible by both")
elif (num % 3==0 and num % 5 != 0):
    print("Divisible by three")
elif (num % 3 !=0 and num % 5 == 0):
    print("Divisible by five")
else:
    print("not divisible by either")
    
```

```python
"""7.ATM Cash Withdrawal

Write a Python program for an ATM.

Rules:

Ask the user to enter:
Account balance
Withdrawal amount
Print:
"Transaction Successful" if:
The withdrawal amount is less than or equal to the balance, and
The withdrawal amount is a multiple of 100.
"Insufficient Balance" if the withdrawal amount is greater than the balance.
"Enter amount in multiples of 100" if the amount is not divisible by 100."""

balance = int(input("enter the balance amount:" ))
withdraw = int(input("enter thewithdrawal amount:"))

if (withdraw <= balance) and (withdraw % 100 == 0) :
    print("Transaction Successful ")
elif withdraw > balance:
    print("Insufficient Balance ")
else:
    print(" Enter amount in multiples of 100")
```

```python
""" 8. Electricity Bill Calculator

Write a program that calculates the electricity bill.

Rules:

Input the number of units consumed.

Charges:

First 100 units → ₹0 per unit
Next 100 units (101–200) → ₹5 per unit
Above 200 units → ₹10 per unit"""


bill = 0
unit = int(input("enter the number of units :"))
if unit <= 100:
    bill=0
    print("total bill=",bill)
elif unit >=101 and unit <= 200:
    bill=((unit-100)*5)
    print("total bill=",bill)
else:
    bill=((unit-200)*10)+(100*5)
    print("total bill=",bill)
```
